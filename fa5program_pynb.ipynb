```python
def calculate_total_price(price_per_item, quantity):
  """
  Calculates the total price of items.

  Args:
      price_per_item (float): The price of one item.
      quantity (int): The number of items purchased.

  Returns:
  float: The total price.
  """
  return price_per_item * quantity

price_input = input("Enter the price of one item: ")
quantity_input = input("Enter the number of items purchased: ")

try:
  price = float(price_input)
  quantity = int(quantity_input)
except ValueError:
  print("Invalid input. Please enter valid numbers for price and quantity.")
else:
  total_price = calculate_total_price(price, quantity)

  print(f"The total price for {quantity} items at ${price:.2f} each is: ${total_price:.2f}")
```
